## Exercice 2

```python
emissions = pd.read_csv("https://koumoul.com/s/data-fair/api/v1/datasets/igt-pouvoir-de-rechauffement-global/convert")
emissions.head(2)
```

```python
# Création

emissions_copy= emissions.loc[:, ['INSEE commune', 'Commune', 'Autres transports', 'Autres transports international']]
emissions_copy.columns
```

```python
# Renommons les variables
emissions_copy= emissions_copy.rename({'INSEE commune': 'code_insee','Autres transports' : 'transports', 'Autres transports international' : 'transports_international' }, axis = 1)

emissions_copy.dtypes
```

Question 3 :

On propose, pour simplifier, de remplacer les valeurs manquantes (NA) par la valeur 0. Utiliser la méthode fillna pour transformer les valeurs manquantes en 0.


```python
emissions_copy = emissions_copy.fillna(0)
```

Question 4. 
On propose, pour simplifier, de remplacer les valeurs manquantes (NA) par la valeur 0. Utiliser la méthode fillna pour transformer les valeurs manquantes en 0.

```python
emissions_copy['dep'] = emissions_copy['code_insee'].str[:2]

emissions_copy[['dep', 'code_insee']].head()
```

```python
emissions_copy['transports_total'] = emissions_copy['transports'] + emissions_copy['transports_international']
emissions_copy['transports_total'].head()
```

Question 5.
Ordonner les données du plus gros pollueur au plus petit puis ordonner les données du plus gros pollueur au plus petit par département (du 01 au 95)

```python
emissions_copy.sort_values(ascending=False, by='transports_total').head()
```

```python
emissions_copy.sort_values('transports_total',ascending=False).groupby(by='dep').head()
```

Question 6. Ne conserver que les communes appartenant aux départements 13 ou 31. Ordonner ces communes du plus gros pollueur au plus petit.

```python
emissions_q6 = emissions_copy.where((emissions_copy['dep'] == '13') | (emissions_copy['dep'] == '31'))
emissions_q6.head(6)
```
